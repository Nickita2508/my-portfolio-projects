```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
import matplotlib.ticker as ticker
import seaborn as sns
import datetime

import warnings
#warnings.filterwarnings("ignore")
sns.set_style('whitegrid')
```

```python
# Загрузка датасетов
df_courses = pd.read_csv('courses.csv', index_col=0)
df_students = pd.read_csv('students.csv')
df_course_contents = pd.read_csv('course_contents.csv')
df_progresses = pd.read_csv('progresses.csv')
df_progress_phases = pd.read_csv('progress_phases.csv')
```

```python
# Объединение датасетов
main = pd.merge(df_progress_phases, df_progresses, how = 'left', left_on = 'progress_id', right_on = 'id')
main = main.drop('id', axis=1)
main = main.merge(df_courses, how = 'left', left_on = 'course_id', right_on = 'id')
main = main.drop('id', axis=1)
main = pd.merge(main, df_course_contents, how = 'left', on = ['course_id', 'module_number', 'lesson_number'])
main.head(3)
```

```python
# Общая информация по итоговому датасету
main.info()
```

### 1.1 Общее количество курсов в датасете

```python
print('\033[1m' + f'Общее количество курсов в датасете: {main['course_id'].nunique()}')
```

### 1.2 Количество модулей на каждом курсе

```python
print('\033[1m' + 'Количество модулей на каждом курсе'.center(100) + '\n')
data_1 = pd.DataFrame(main.groupby('title')['module_number'].nunique().sort_values(ascending=False))
data_1.index.name, data_1.columns  = 'Название курса', ['Количество модулей']
data_1
```

### 1.6 Минимальный, максимальный, средний, медианный возраст студентов

```python
# Изучим информацию о датасете "students"
df_students.info()
```

```python
df_students_filtered = df_students.copy()
# Приведем столбец "birthday" к типу datetime
df_students_filtered['birthday'] = pd.to_datetime(df_students_filtered['birthday'], errors = 'coerce')
# Избавимся от 'Na' значений
df_students_filtered = df_students_filtered.dropna(subset=['birthday'])
# Создадим дополнительный столбец для возраста
df_students_filtered['age'] = ((datetime.datetime.now() - df_students_filtered['birthday']).dt.days // 365).astype(int)
# Посмотрим на распределение студентов по возрастам, наличие выбросов
sns.displot(df_students_filtered['age'], kde=True)
plt.title('Распределение студентов по возрастам');
```

```python
# Избавимся от аномалий в данных по правилу трех сигм
# Наблюдаемое отклонение от среднего
dev = (df_students_filtered['age'] - df_students_filtered['age'].mean()).abs()
# Стандартное отклонение
std = df_students_filtered['age'].std()
# Фильтруем исходный набор данных
df_students_filtered = df_students_filtered[~(dev>3*std)]
# Посмотрим на распределение студентов по возрастам без аномалий
sns.displot(df_students_filtered['age'], kde=True)
plt.title('Распределение студентов по возрастам без отклонений');
```

```python
print('\033[1m' + f'Минимальный возраст студентов: {df_students_filtered['age'].min()} лет')
print('\033[1m' + f'Максимальный возраст студентов: {df_students_filtered['age'].max()} лет')
print('\033[1m' + f'Средний возраст студентов: {df_students_filtered['age'].mean():.0f} года')
print('\033[1m' + f'Медианный возраст студентов: {df_students_filtered['age'].median():.0f} года')
```

### 1.7 Минимальный, максимальный, средний, медианный возраст студентов на каждом курсе

```python
age_students_courses = pd.merge(main[['student_id', 'title']], df_students_filtered[['id', 'age']], how='left', left_on='student_id', right_on='id')
age_students_courses = age_students_courses.drop('id',axis=1)
age_students_courses.columns = ['student_id', 'Название курса', 'Возраст']
data_4 = age_students_courses.drop_duplicates()
data_4 = data_4.groupby('Название курса').agg({'Возраст' : ['min', 'max', 'mean', 'median']}).map(int)
print('\033[1m' + 'Минимальный, максимальный, средний, медианный возраст студентов на каждом курсе')
data_4
```

### 2. Bar-chart, отражающий количество студентов на каждом курсе

```python
# Подготовим данные
students_count = main[['title', 'student_id']].drop_duplicates().groupby('title').count()
students_count = students_count.merge(df_courses, how='left', on='title').drop('id', axis=1)
students_count.columns = ['Курс', 'Количество студентов', 'Сфера']
students_count = students_count.sort_values('Количество студентов', ascending=False)
# Строим график
plt.figure(figsize=(15,6), dpi=100)
sns.barplot(data=students_count, x='Курс', y='Количество студентов', hue='Сфера', palette='husl')
plt.title('Количество студентов на каждом курсе', fontsize=16)
plt.xticks(rotation=45, ha='right');
```

### 3. Горизонтальный bar-chart, отражающий количество студентов на каждом курсе

```python
fig, ax = plt.subplots(figsize=(15,6), dpi=100)
sns.barplot(data=students_count, y='Курс', x='Количество студентов', hue='Сфера', palette='husl', orient='y', alpha=0.5)
ax.bar_label(ax.containers[0])
ax.axvline(x=students_count['Количество студентов'].median(), linewidth=1, color='b', alpha=0.5, ls='--')
ax.set_title('Количество студентов на каждом курсе', fontsize=16)
ax.set_ylabel(None)
plt.box(False);
```

Если домашних заданий в модуле несколько, то считать разность между временем начала выполнения первой домашней работы и временем окончания выполнения последней домашней работы в модуле. ### 1. Минимальное, максимальное, среднее, медианное время прохождения каждого модуля

```python
# Для начала создадим новый датафрейм в котором не будет записей без финишной даты и удалим столбцы, которые нам не требуются
data_9 = main[main['is_homework']==True].dropna(subset=['finish_date'], axis=0).drop(['lesson_title','is_video', 'is_homework','status'], axis=1)

# Чтобы вычислить разность между временем начала выполнения первой домашней работы и временем окончания выполнения последней домашней работы в модуле
# создадим два датафрейма,в первом будут строки соответсвующие первому ДЗ в модуле и start_date
data_9_start = (data_9
          .groupby(['progress_id', 'student_id', 'course_id', 'title', 'field', 'module_title', 'module_number'])
          .apply(lambda x: x.loc[x['lesson_number'].idxmin()], include_groups=False)
          [['lesson_number', 'lesson_token','start_date']]
          .reset_index())

# во втором будут строки соответсвующие последнему ДЗ в модуле и finish_date
data_9_finish = (data_9
          .groupby(['progress_id', 'student_id', 'course_id', 'title', 'field', 'module_title', 'module_number'])
          .apply(lambda x: x.loc[x['lesson_number'].idxmax()], include_groups=False)
          [['lesson_number', 'lesson_token','finish_date']]
          .reset_index())

# Объединим полученные датафреймы в один
data_9_full = pd.merge(data_9_start,
                data_9_finish,
                how='inner',
                on=['progress_id', 'student_id', 'course_id', 'title', 'field', 'module_title', 'module_number'],
                suffixes=['_start', '_finish'])

# Приведем колонки с датами к типу datetime
data_9_full['start_date'] = pd.to_datetime(data_9_full['start_date'].apply(lambda x: x[:19]))
data_9_full['finish_date'] = pd.to_datetime(data_9_full['finish_date'].apply(lambda x: x[:19]))

# Вычислим время прохождения модуля в днях
data_9_full['diff_days'] =  (data_9_full['finish_date'] - data_9_full['start_date']) / np.timedelta64 (1 , 'D')
# Исключим ошибки в датах
data_9_full = data_9_full[data_9_full['diff_days'] > 0]

# создадим таблицу с минимальным, максимальным, средним, медианным временем прохождения каждого модуля 
data_9_stats = (data_9_full
    .groupby(['title', 'module_number', 'module_title'])
    .agg({'diff_days': ['min', 'max', 'mean', 'median']})['diff_days'].round(2).reset_index())

print('\033[1m' + 'Минимальное, максимальное, среднее, медианное время прохождения каждого модуля'.center(150) + '\n')
data_9_stats
```

### 2. line-graph с медианным временем прохождения каждого модуля для каждого курса

```python
fig, axes = plt.subplots(nrows=5, ncols=3, figsize=(32,32), dpi=100)
# С помощью цикла для каждого графика добавляем данные, выбираем его тип, подписываем название графиков и подписываем оси
num = 0
for i in range(3):
    for j in range(5):
        ax = axes[j][i]
        df_plot = data_9_stats[data_9_stats['title'] == courses[num]].sort_values('module_number')
        df_plot['module'] =  df_plot['module_number'].astype(str) + '. ' +  df_plot['module_title'].apply(lambda x: f'{x[:20]}...')
        sns.lineplot(x='module', y='median', data=df_plot, ax=ax)
        ax.set_title(f'{courses[num]}')
        num += 1
        ax.tick_params(axis='x', labelrotation=45)
        if i == 0:
            ax.set_ylabel('Кол-во дней')
        else:
            ax.set_ylabel(None)
        if j == 4:  
            ax.set_xlabel('Кол-во дней')
        else:
            ax.set_xlabel(None)
fig.suptitle('Медианное время прохождения каждого модуля для каждого курса', y=1, fontsize=20)
plt.tight_layout()
plt.show()
```

### 3. Cезонность. Медианное время выполнения домашней работы по месяцам для каждого курса

```python
# Выбираем данные
data_10 = main[main['is_homework']==True].dropna(subset=['finish_date'], axis=0)[['start_date', 'finish_date', 'title']].copy()

# Преобразуем данные в datetime
data_10['start_date'] = pd.to_datetime(data_10['start_date'].apply(lambda x: x[:19]))
data_10['finish_date'] = pd.to_datetime(data_10['finish_date'].apply(lambda x: x[:19]))
data_10['diff_days'] =  (data_10['finish_date'] - data_10['start_date']) / np.timedelta64 (1 , 'D')

# Добавляем столбец с номером месяца
data_10['month'] =  data_10['start_date'].dt.month

# Добавляем столбец с названием месяца
data_10['month_name'] = data_10['start_date'].dt.month_name(locale='Russian')

# Выводим таблицу
data_10_unstacked = data_10.groupby(['title', 'month', 'month_name']).agg({'diff_days': 'median'}).unstack(0, fill_value=0).reset_index()
data_10_unstacked = data_10_unstacked.drop(['month'], axis=1).set_index('month_name')
print('\033[1m' + 'Медианное время выполнения домашней работы по месяцам для каждого курса'.center(150) + '\n')
data_10_unstacked
```

## 4. Расчет конверсии

1. Посчитать конверсию перехода студентов из одного модуля в другой на каждом курсе. Формула: отношение количества студентов, приступивших к выполнению домашнего задания в этом модуле (если ДЗ в модуле несколько, то считать по первому ДЗ в модуле), к количеству студентов, сдавших задание в предыдущем модуле (если ДЗ в модуле несколько, то считать по последнему ДЗ в модуле).

Графики должны иметь заголовки. Ticks должны содержать номер и название модуля. Цвет столбцов графиков должен содержать информацию о сфере, к которой относится курс (нужна легенда). Прозрачность должна стоять на отметке 0.1. На графики должна быть нанесена линия медианы конверсии для каждого курса. У медианы должен быть свой цвет. Рамок у графиков быть не должно <br>

### 1. Конверсия перехода студентов из одного модуля в другой на каждом курсе

```python
# отберем данные
data_11 = (main[(main['is_homework']==True)].groupby(['title', 'module_title', 'student_id', 'module_number', 'status', 'field'])
    .agg({'start_date': 'first', 'finish_date': 'last'})
    .reset_index())
# количество студентов начавших ДЗ
data_11_start = data_11.groupby(['title', 'module_number', 'module_title', 'field'])['student_id'].count().reset_index(name='students_start')
# количество студентов закончивших ДЗ
data_11_done = data_11[data_11['status'] == 'done'].groupby(['title', 'module_number', 'module_title', 'field'])['student_id'].count().reset_index(name='students_done')
# соединим датафреймы
data_11_full = pd.merge(data_11_start, data_11_done, how='inner')
# Рассчитаем отношение студентов начавших ДЗ к студентам закочавших последнее ДЗ в предыдущем модуле
data_11_full['conversion'] = round(data_11_full['students_start'] / (data_11_full['students_done'] - data_11_full.groupby('title')['students_done'].diff().fillna(np.nan)), 3).fillna(0)                                            
data_11_pivot = pd.pivot(data=data_11_full, index='title', columns='module_number', values='conversion').map(lambda x: str(x) if pd.notna(x) else '')
print('\033[1m' + 'Конверсия перехода студентов из одного модуля в другой на каждом курсе'.center(150) + '\n')
data_11_pivot
```

### 2. bar-chart, отражающий конверсию перехода студентов из одного модуля в другой на каждом курсе

```python
fig, ax = plt.subplots(figsize=(30,10))
sns.barplot(data=data_11_full, x='title', y='conversion', hue='module_number')
ax.tick_params(axis='x', labelrotation=45)
# легенда
plt.xlabel('Курсы')
plt.ylabel('Конверсия')
fig.suptitle('Конверсия перехода студентов из одного модуля в другой на каждом курсе', y=0.92, size=20)
plt.ylim(00,1.5)
plt.box(False)

plt.show()
```

### 3. Горизонтальный bar-chart, отражающий конверсию перехода студентов из одного модуля в другой на каждом курсе

```python
fig, axes = plt.subplots(nrows=15, ncols=1, figsize=(10,100))
for i, course in enumerate(courses):
    df_plot = data_11_full[data_11_full['title'] == course].sort_values('module_number')
    df_plot['module'] =  df_plot['module_number'].astype(str) + '. ' +  df_plot['module_title'].apply(lambda x: f'{x[:25]}...')
    ax = axes[i]
    sns.barplot(x='conversion', y='module', data=df_plot, ax=ax, orient='y',hue='module_number', legend=False, palette='rocket_r')
    ax.set_title(f'{course}')
    ax.set_ylabel(None)
    ax.set_xlabel('Конверсия')
fig.suptitle('Конверсия перехода студентов из одного модуля в другой на каждом курсе', y=0.89, fontsize=20)
plt.show()
```
